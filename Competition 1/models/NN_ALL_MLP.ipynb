````python
import os
import sys
import warnings
warnings.filterwarnings('ignore')

# Add parent directory to path to import from helpers
current_dir = os.getcwd()
parent_dir = os.path.abspath(os.path.join(current_dir, '..'))

if parent_dir not in sys.path:
    sys.path.insert(0, parent_dir)

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

# Data preprocessing
from helpers.process_data import preprocess_data, in_fold_process_data
from helpers.nlp import TextFeatureExtractor, combine_text_columns

# Scikit-learn
from sklearn.model_selection import StratifiedKFold
from sklearn.metrics import accuracy_score, roc_auc_score, classification_report, confusion_matrix, roc_curve, auc
from sklearn.preprocessing import StandardScaler

# Neural Network
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, callbacks, regularizers
from tensorflow.keras.optimizers import Adam

# Set random seeds for reproducibility
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)
tf.random.set_seed(RANDOM_STATE)

print("✅ All libraries imported successfully!")
print(f"TensorFlow version: {tf.__version__}")
print(f"GPU available: {len(tf.config.list_physical_devices('GPU')) > 0}")
````

````python
# Load data
print("="*70)
print("LOADING DATA")
print("="*70)

train = pd.read_csv('../data/train.csv')
test = pd.read_csv('../data/test.csv')

# Separate features and target
y_train = train['target'].copy()

print(f"\n📊 Dataset Summary:")
print(f"  Training samples: {len(train):,}")
print(f"  Test samples: {len(test):,}")
print(f"  Features: {train.shape[1] - 1}")

print(f"\n🎯 Target Distribution:")
target_counts = y_train.value_counts().sort_index()
for label, count in target_counts.items():
    percentage = (count / len(y_train)) * 100
    emoji = "🤖" if label == 1 else "👤"
    print(f"  {emoji} Class {label}: {count:,} samples ({percentage:.1f}%)")

print("\n✅ Data loaded successfully!")
````

````python
# Extract both TEXT and STRUCTURED features
print("="*70)
print("COMBINED FEATURE EXTRACTION (TEXT + STRUCTURED)")
print("="*70)

# Text columns to process
TEXT_COLUMNS = ['location', 'description', 'screen_name']

# Separate features and target
raw_train_features = train.drop('target', axis=1)

print(f"\n{'='*70}")
print("PART 1: TEXT FEATURE EXTRACTION")
print(f"{'='*70}")

# Combine text columns
print(f"\n📝 Combining text columns: {TEXT_COLUMNS}")
train_text = combine_text_columns(train, TEXT_COLUMNS)
test_text = combine_text_columns(test, TEXT_COLUMNS)

print(f"\n✅ Text combined successfully!")
print(f"  Sample combined text: '{train_text.iloc[0][:100]}...'")

# Extract text features using TF-IDF + SVD
print(f"\n🔧 Extracting text features using TF-IDF + SVD...")
print(f"  Word-level n-grams: (1, 2)")
print(f"  Char-level n-grams: (3, 5)")
print(f"  SVD components: 300")

text_extractor = TextFeatureExtractor(n_components=300, random_state=RANDOM_STATE)
X_text_train = text_extractor.fit_transform(train_text)
X_text_test = text_extractor.transform(test_text)

# Convert to DataFrames with proper column names
text_feature_names = [f'text_feat_{i}' for i in range(X_text_train.shape[1])]
X_text_train_df = pd.DataFrame(X_text_train, columns=text_feature_names, index=train.index)
X_text_test_df = pd.DataFrame(X_text_test, columns=text_feature_names, index=test.index)

print(f"\n✅ Text features extracted successfully!")
print(f"  Text feature shape - Train: {X_text_train_df.shape}, Test: {X_text_test_df.shape}")

print(f"\n{'='*70}")
print("PART 2: STRUCTURED FEATURE EXTRACTION")
print(f"{'='*70}")

print(f"\n📝 Processing structured features using process_data pipeline...")
print(f"  This includes numerical, binary, categorical, and engineered features")

# Step 1: Pre-CV preprocessing (deterministic transformations)
print("\n1. Pre-CV preprocessing (deterministic transformations)...")
X_struct_preprocessed = preprocess_data(raw_train_features, verbose=True, include_statuses_count=True)
X_struct_test_preprocessed = preprocess_data(test, verbose=False, include_statuses_count=True)

print(f"\n✅ Structured features preprocessed:")
print(f"  Train: {X_struct_preprocessed.shape}")
print(f"  Test: {X_struct_test_preprocessed.shape}")

# Step 2: Get feature names from in-fold processing (sample)
print(f"\n2. Getting structured feature names from in-fold processing...")
print(f"   (Using first 1000 samples as dummy fold)")

dummy_train_idx = range(min(1000, len(X_struct_preprocessed)))
dummy_val_idx = range(min(1000, len(X_struct_preprocessed)), min(2000, len(X_struct_preprocessed)))

X_dummy_train = X_struct_preprocessed.iloc[dummy_train_idx]
X_dummy_val = X_struct_preprocessed.iloc[dummy_val_idx]

# Process to get final feature set
X_struct_sample, _, _ = in_fold_process_data(
    X_dummy_train, X_dummy_val,
    imputation_strategy='median',
    categorical_encoding='onehot',
    lang_frequency_threshold=0,
    scale_numerical=False,  # We'll scale separately with StandardScaler
    verbose=True
)

n_struct_features = X_struct_sample.shape[1]
struct_feature_names = X_struct_sample.columns.tolist()

print(f"\n✅ Structured features extracted successfully!")
print(f"  Total structured features: {n_struct_features}")
print(f"  Feature categories:")
print(f"    - Numerical features (raw + ratios + per-day metrics)")
print(f"    - Binary features (4 features)")
print(f"    - Categorical features (one-hot encoded languages)")
print(f"    - Text-derived features (lengths, patterns)")
print(f"    - Engineered features (profile quality, interaction ratios)")

print(f"\n{'='*70}")
print("COMBINED FEATURES SUMMARY")
print(f"{'='*70}")
print(f"\n📊 Feature Dimensions:")
print(f"  Text features: {X_text_train_df.shape[1]}")
print(f"  Structured features: ~{n_struct_features} (after one-hot encoding)")
print(f"  Total combined features: ~{X_text_train_df.shape[1] + n_struct_features}")

print(f"\n💡 These features will be combined and processed in each CV fold")
print(f"   to avoid data leakage while maintaining all information")
````

# Neural Network Configuration

````python
# =============================================================================
# NEURAL NETWORK CONFIGURATION
# =============================================================================

# Model architecture
N_HIDDEN_LAYERS = 4
HIDDEN_UNITS = [512, 256, 128, 64]
DROPOUT_RATE = 0.35
L2_REG = 0.001

# Training configuration
BATCH_SIZE = 256
EPOCHS = 150
LEARNING_RATE = 0.001
EARLY_STOPPING_PATIENCE = 20

# Cross-validation
N_SPLITS = 5

# Estimate combined features
n_text_features = X_text_train_df.shape[1]
n_combined_features = n_text_features + n_struct_features

print("="*70)
print("NEURAL NETWORK CONFIGURATION")
print("="*70)
print(f"\n🏗️ Architecture:")
print(f"  Input features: ~{n_combined_features} (text + structured)")
print(f"    - Text features: {n_text_features}")
print(f"    - Structured features: ~{n_struct_features}")
print(f"  Hidden layers: {N_HIDDEN_LAYERS}")
print(f"  Hidden units: {HIDDEN_UNITS}")
print(f"  Dropout rate: {DROPOUT_RATE}")
print(f"  L2 regularization: {L2_REG}")

print(f"\n⚙️ Training:")
print(f"  Batch size: {BATCH_SIZE}")
print(f"  Max epochs: {EPOCHS}")
print(f"  Learning rate: {LEARNING_RATE}")
print(f"  Early stopping patience: {EARLY_STOPPING_PATIENCE}")

print(f"\n🔄 Cross-validation:")
print(f"  Folds: {N_SPLITS}")

print(f"\n💡 Note: Combining TEXT + STRUCTURED features for maximum performance")
print(f"  This leverages both textual patterns and account behavior patterns")
print("="*70)
````

````python
# Build Neural Network Model
def build_mlp_model(input_dim, hidden_units=HIDDEN_UNITS, dropout_rate=DROPOUT_RATE, l2_reg=L2_REG, learning_rate=LEARNING_RATE):
    """
    Build a Multi-Layer Perceptron (MLP) model for binary classification.
    
    Parameters:
    -----------
    input_dim : int
        Number of input features (combined text + structured)
    hidden_units : list
        List of hidden layer sizes
    dropout_rate : float
        Dropout rate for regularization
    l2_reg : float
        L2 regularization strength
    learning_rate : float
        Learning rate for optimizer
    
    Returns:
    --------
    model : keras.Model
        Compiled Keras model
    """
    model = keras.Sequential(name='MLP_Combined_Classifier')
    
    # Input layer
    model.add(layers.Input(shape=(input_dim,), name='combined_input'))
    
    # Hidden layers with batch normalization and dropout
    for i, units in enumerate(hidden_units):
        model.add(layers.Dense(
            units, 
            activation='relu',
            kernel_regularizer=regularizers.l2(l2_reg),
            name=f'hidden_{i+1}'
        ))
        model.add(layers.BatchNormalization(name=f'bn_{i+1}'))
        model.add(layers.Dropout(dropout_rate, name=f'dropout_{i+1}'))
    
    # Output layer
    model.add(layers.Dense(1, activation='sigmoid', name='output'))
    
    # Compile model
    model.compile(
        optimizer=Adam(learning_rate=learning_rate),
        loss='binary_crossentropy',
        metrics=[
            'accuracy',
            keras.metrics.AUC(name='auc'),
            keras.metrics.Precision(name='precision'),
            keras.metrics.Recall(name='recall')
        ]
    )
    
    return model

# Build and display model architecture
print("="*70)
print("BUILDING NEURAL NETWORK MODEL")
print("="*70)

# Use estimated combined features for model summary
model = build_mlp_model(input_dim=n_combined_features)

print("\n📋 Model Summary:")
model.summary()

print(f"\n✅ Model built successfully!")
print(f"  Total parameters: {model.count_params():,}")
print(f"  Input: Text features ({n_text_features}) + Structured features (~{n_struct_features})")
````

````python
# Train MLP with 5-fold Cross Validation (Combined Features)
print("\n" + "="*70)
print("TRAINING MLP WITH 5-FOLD CROSS-VALIDATION")
print("Combined Text + Structured Features")
print("="*70)

# Setup cross-validation
skf = StratifiedKFold(n_splits=N_SPLITS, shuffle=True, random_state=RANDOM_STATE)

# Storage for results
oof_predictions = np.zeros(len(y_train))
test_predictions_folds = []
cv_scores = []
fold_histories = []

print(f"\n🔄 Processing features within each CV fold to avoid data leakage...")

# Cross-validation loop
for fold, (train_idx, val_idx) in enumerate(skf.split(X_struct_preprocessed, y_train), 1):
    print(f"\n{'='*20} FOLD {fold}/{N_SPLITS} {'='*20}")
    
    # Split preprocessed structured data
    X_struct_train_fold = X_struct_preprocessed.iloc[train_idx]
    X_struct_val_fold = X_struct_preprocessed.iloc[val_idx]
    
    # Split text features
    X_text_train_fold = X_text_train_df.iloc[train_idx]
    X_text_val_fold = X_text_train_df.iloc[val_idx]
    
    # Split target
    y_train_fold = y_train.iloc[train_idx].values
    y_val_fold = y_train.iloc[val_idx].values
    
    print(f"  Structured - Train: {X_struct_train_fold.shape}, Val: {X_struct_val_fold.shape}")
    print(f"  Text - Train: {X_text_train_fold.shape}, Val: {X_text_val_fold.shape}")
    
    # In-fold processing for structured features (imputation, encoding, etc.)
    print(f"  Performing in-fold processing (imputation & encoding)...")
    X_struct_train_processed, X_struct_val_processed, _ = in_fold_process_data(
        X_struct_train_fold, X_struct_val_fold,
        imputation_strategy='median',
        categorical_encoding='onehot',
        lang_frequency_threshold=0,
        scale_numerical=False,  # We'll scale separately
        verbose=False
    )
    
    print(f"  Processed structured - Train: {X_struct_train_processed.shape}, Val: {X_struct_val_processed.shape}")
    
    # Combine text and structured features
    X_train_combined = pd.concat([X_text_train_fold.reset_index(drop=True), 
                                   X_struct_train_processed.reset_index(drop=True)], axis=1)
    X_val_combined = pd.concat([X_text_val_fold.reset_index(drop=True), 
                                 X_struct_val_processed.reset_index(drop=True)], axis=1)
    
    print(f"  Combined features - Train: {X_train_combined.shape}, Val: {X_val_combined.shape}")
    
    # Apply StandardScaler (fit on train, transform both)
    scaler_fold = StandardScaler()
    X_train_scaled = scaler_fold.fit_transform(X_train_combined)
    X_val_scaled = scaler_fold.transform(X_val_combined)
    
    print(f"  Scaled features - Mean: {X_train_scaled.mean():.6f}, Std: {X_train_scaled.std():.6f}")
    
    # Process test data for this fold
    _, X_struct_test_processed, _ = in_fold_process_data(
        X_struct_train_processed.iloc[:100],
        X_struct_test_preprocessed,
        imputation_strategy='median',
        categorical_encoding='onehot',
        lang_frequency_threshold=0,
        scale_numerical=False,
        verbose=False
    )
    
    # Align test structured features with training structured features
    if X_struct_train_processed.shape[1] != X_struct_test_processed.shape[1]:
        missing_cols = set(X_struct_train_processed.columns) - set(X_struct_test_processed.columns)
        extra_cols = set(X_struct_test_processed.columns) - set(X_struct_train_processed.columns)
        for col in missing_cols:
            X_struct_test_processed[col] = 0
        for col in extra_cols:
            X_struct_test_processed = X_struct_test_processed.drop(columns=[col])
        X_struct_test_processed = X_struct_test_processed[X_struct_train_processed.columns]
    
    # Combine test features
    X_test_combined = pd.concat([X_text_test_df.reset_index(drop=True), 
                                  X_struct_test_processed.reset_index(drop=True)], axis=1)
    
    X_test_scaled = scaler_fold.transform(X_test_combined)
    
    # Build model for this fold
    model_fold = build_mlp_model(
        input_dim=X_train_scaled.shape[1],
        hidden_units=HIDDEN_UNITS,
        dropout_rate=DROPOUT_RATE,
        l2_reg=L2_REG,
        learning_rate=LEARNING_RATE
    )
    
    # Callbacks
    early_stop = callbacks.EarlyStopping(
        monitor='val_auc',
        patience=EARLY_STOPPING_PATIENCE,
        restore_best_weights=True,
        mode='max',
        verbose=0
    )
    
    reduce_lr = callbacks.ReduceLROnPlateau(
        monitor='val_auc',
        factor=0.5,
        patience=7,
        min_lr=1e-7,
        mode='max',
        verbose=0
    )
    
    # Train model
    print(f"  Training model...")
    history = model_fold.fit(
        X_train_scaled, y_train_fold,
        validation_data=(X_val_scaled, y_val_fold),
        epochs=EPOCHS,
        batch_size=BATCH_SIZE,
        callbacks=[early_stop, reduce_lr],
        verbose=0
    )
    
    fold_histories.append(history)
    
    # Predict on validation fold
    val_pred = model_fold.predict(X_val_scaled, verbose=0).flatten()
    oof_predictions[val_idx] = val_pred
    
    # Predict on test set
    test_pred = model_fold.predict(X_test_scaled, verbose=0).flatten()
    test_predictions_folds.append(test_pred)
    
    # Calculate fold metrics
    fold_auc = roc_auc_score(y_val_fold, val_pred)
    fold_acc = accuracy_score(y_val_fold, (val_pred > 0.5).astype(int))
    cv_scores.append(fold_auc)
    
    # Get best epoch
    best_epoch = np.argmax(history.history['val_auc']) + 1
    best_val_auc = max(history.history['val_auc'])
    
    print(f"  Best epoch: {best_epoch}/{len(history.history['loss'])}")
    print(f"  Validation AUC: {fold_auc:.4f}")
    print(f"  Validation Accuracy: {fold_acc:.4f}")

# Average test predictions across folds
test_predictions = np.mean(test_predictions_folds, axis=0)

# Overall CV performance
overall_auc = roc_auc_score(y_train, oof_predictions)
overall_acc = accuracy_score(y_train, (oof_predictions > 0.5).astype(int))

print(f"\n{'='*70}")
print("CROSS-VALIDATION RESULTS")
print(f"{'='*70}")
print(f"\n📊 Overall Performance:")
print(f"  Mean CV AUC: {np.mean(cv_scores):.4f} (+/- {np.std(cv_scores):.4f})")
print(f"  OOF AUC: {overall_auc:.4f}")
print(f"  OOF Accuracy: {overall_acc:.4f}")
print(f"\n📈 Individual Fold AUCs:")
for i, score in enumerate(cv_scores, 1):
    print(f"  Fold {i}: {score:.4f}")

print(f"\n{'='*70}")
````

## 📝 Notes and Usage Instructions

### About This Notebook

This notebook implements a **Multi-Layer Perceptron (MLP) Neural Network** for Twitter bot detection using **COMBINED text and structured features**:

**Text Features** (300 features via TF-IDF + SVD):
- **location**: User's location text
- **description**: User's bio/description
- **screen_name**: Username

**Structured Features** (~60+ features):
- **Numerical**: followers_count, friends_count, statuses_count, etc.
- **Binary**: default_profile, default_profile_image, verified, geo_enabled
- **Categorical**: lang (with one-hot encoding)
- **Engineered**: ratios, interactions, account age features

### Key Features

✅ **Combined Feature Set** - Leverages both textual patterns and account behavior  
✅ **Text Processing** - TF-IDF (word 1-2 grams, char 3-5 grams) + TruncatedSVD  
✅ **Structured Processing** - Two-stage preprocessing prevents data leakage  
✅ **Deep Neural Network** - 4 hidden layers (512, 256, 128, 64 units)  
✅ **Regularization** - Dropout (0.35) + L2 regularization + Batch normalization  
✅ **5-Fold Cross-Validation** - Robust performance estimation  
✅ **Early Stopping** - Prevents overfitting  

### Architecture Details

**Input Layer:**
- ~360+ combined features (300 text + 60+ structured)

**Hidden Layers:**
1. Dense(512) → BatchNorm → ReLU → Dropout(0.35)
2. Dense(256) → BatchNorm → ReLU → Dropout(0.35)
3. Dense(128) → BatchNorm → ReLU → Dropout(0.35)
4. Dense(64) → BatchNorm → ReLU → Dropout(0.35)

**Output Layer:**
- Dense(1) → Sigmoid (binary classification)
